# Simple authorize


````python
# check token if available
if 'authorize' in globals():
    print("Already authorized")
    await session.get("https://api.hh.ru/me", headers={
        "Authorization" : f"{token_type} {access_token}",
        "HH-User-Agent" : hh_useragent,
    })

# take new token
else:
    authorize = await session.post("https://api.hh.ru/token", 
        headers={
            "Content-Type" : "application/x-www-form-urlencoded",
            "User-Agent" : hh_useragent
        },
        data={
            "client_id" : hh_clientid,
            "client_secret" : hh_secret,
            "grant_type" : "client_credentials"
        }
    )
print(await authorize.text())
print(json.dumps(dict(authorize.headers.items()), indent=4))
data =  await authorize.json()
access_token = data["access_token"]
token_type = str(data["token_type"]).capitalize()
````

# Fetch jobs order by date

````python
page = 0
per_page = 100
text = "Python"
salary = 50000

jobs = await session.get(f"https://api.hh.ru/vacancies",
    params={
        "page" : page,
        "per_page" : per_page,
        "text" : text,
        "salary" : salary,
        #"order_by" : "publication_time"
        "date_from" : str(datetime.datetime.now(datetime.timezone.utc).isoformat())
    },
    headers={
        "Authorization" : f"{token_type} {access_token}"
    }
)
jobs_data = await jobs.json()
print(json.dumps(jobs_data, indent=4, ensure_ascii=False))
````

# Parsing data

````python
parsed = []
for i in jobs_data['items']:
    salary = i.get('salary') or {}
    parsed.append({
            'id': i.get('id'),
            'area': i.get('area', {}).get('name'),
            'employer': i.get('employer', {}).get('name'),
            'salary_from': salary.get('from'),
            'salary_to': salary.get('to'),
            'currency': salary.get('currency'),
            'gross': salary.get('gross')
        })
parsed = pd.DataFrame(parsed)
parsed
````

````python
def calculate_avg_salary(row):
    salary_from = row['salary_from']
    salary_to = row['salary_to']
    
    if pd.notna(salary_from) and pd.notna(salary_to):
        return (salary_from + salary_to) / 2
    elif pd.notna(salary_from):
        return salary_from
    elif pd.notna(salary_to):
        return salary_to
    else:
        return None

parsed['avg_salary'] = parsed.apply(calculate_avg_salary, axis=1)
df_with_salary = parsed[
    (parsed['currency'] == 'RUR') & 
    (parsed['avg_salary'].notna())
]

print("\n" + "="*50)
print("СТАТИСТИКА ПО ЗАРПЛАТАМ (RUR)")
print("="*50)

median_salary = df_with_salary['avg_salary'].median()
mean_salary = df_with_salary['avg_salary'].mean()
min_salary = df_with_salary['avg_salary'].min()
max_salary = df_with_salary['avg_salary'].max()

print(f"Медиана: {median_salary:,.0f} ₽")
print(f"Среднее: {mean_salary:,.0f} ₽")
print(f"Минимум: {min_salary:,.0f} ₽")
print(f"Максимум: {max_salary:,.0f} ₽")
print(f"Вакансий с зарплатой: {len(df_with_salary)} из {len(parsed)}")

print("\n" + "="*50)
print("МЕДИАНА ПО ГОРОДАМ")
print("="*50)

salary_by_city = df_with_salary.groupby('area')['avg_salary'].agg([
    ('count', 'count'),
    ('median', 'median'),
    ('mean', 'mean')
]).round(0)
print(salary_by_city)
````
